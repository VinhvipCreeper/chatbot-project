```python
import pandas as pd
```

```python
# Load the 10-K financial data
df = pd.read_csv('Data/10-K_Extract__MSFT_TSLA_AAPL_.csv')

# Ensure data is sorted so pct_change() computes correctly (oldest → newest within each company)
df = df.sort_values(['Company', 'Fiscal Year']).reset_index(drop=True)

print(df.to_string(index=False))
```

## Year-over-Year Growth Analysis

Calculate percentage change for each financial metric grouped by company.

```python
metrics = ['Total Revenue', 'Net Income', 'Total Assets', 'Total Liabilities', 'CFO']

for metric in metrics:
    df[f'{metric} Growth (%)'] = (
        df.groupby('Company')[metric]
        .pct_change() * 100
    ).round(2)

growth_cols = ['Company', 'Fiscal Year'] + [f'{m} Growth (%)' for m in metrics]
print(df[growth_cols].to_string(index=False))
```

```python
# Aggregate: latest-year absolute values and average YoY growth per company
summary = df.groupby('Company').agg(
    Latest_Year=('Fiscal Year', 'max'),
    Latest_Revenue=('Total Revenue', 'last'),
    Latest_Net_Income=('Net Income', 'last'),
    Avg_Revenue_Growth=('Total Revenue Growth (%)', 'mean'),
    Avg_NetIncome_Growth=('Net Income Growth (%)', 'mean'),
    Avg_CFO_Growth=('CFO Growth (%)', 'mean'),
).round(2)

print(summary.to_string())
```

## Summary of Findings

### Apple (AAPL)
- Revenue has been essentially flat (~$390B range), with slight YoY fluctuations (~−2% to +2%).
- Net income is similarly stable (~$94–$100B), reflecting a mature, highly profitable business.
- CFO grew steadily, reaching **$118B** in FY2024, signaling strong cash generation even without top-line growth.

### Microsoft (MSFT)
- Revenue grew at the strongest pace of the three (~15–17% YoY), driven by Azure cloud and AI services, reaching **$282B** in FY2025.
- Net income grew even faster (~18–22% YoY), reflecting significant operating leverage.
- CFO grew from $88B (FY2023) to **$136B** (FY2025) — a ~55% increase in two years.
- Microsoft is the highest-growth large-cap among the three.

### Tesla (TSLA)
- Revenue grew ~20% from FY2022 to FY2023 but slowed significantly to just +1% in FY2024 (~$97.7B), reflecting pricing pressure and market saturation in EVs.
- Net income fell sharply in FY2024 (~−52%), a major reversal after record profits in FY2023.
- CFO remained positive but declined, highlighting margin compression.
- Tesla shows the most volatility and risk among the three.
